```python
# [CELL 3] LINEAR GROWTH
def linear_operations(n):
    count = 0
    for _ in range(n):
        count += 1
    return count

for n in [10, 100, 1000, 5000, 10000]:
    print(f"n={n:5d} | operations={linear_operations(n):5d}")
```

```python
# [CELL 4] BIG O CLASSIFICATION
def log_ops(n):
    count = 0
    while n > 1:
        n //= 2
        count += 1
    return count

for n in [10, 100, 1000]:
    print(
        f"n={n:4d} | O(1)=1 | "
        f"O(log n)={log_ops(n):2d} | "
        f"O(n)={n:4d} | O(n^2)={n*n:7d}"
    )
```

```python
# [CELL 5] SPACE COMPLEXITY
def constant_space(n):
    total = 0
    for value in range(n):
        total += value
    return total

def linear_space(n):
    return list(range(n))

for n in [10, 100, 1000, 5000]:
    data = linear_space(n)
    print(f"n={n:5d} | O(1): constant | O(n): {len(data)}")
```

```python
# [CELL 7] LINEAR SEARCH
def linear_search(data, target):
    for i, value in enumerate(data):
        if value == target:
            return i, i + 1
    return -1, len(data)

data = list(range(1, 21))
target = SEED_DIGIT + 1

index, comparisons = linear_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 8] BINARY SEARCH
def binary_search(data, target):
    low, high, comparisons = 0, len(data) - 1, 0
    while low <= high:
        mid = (low + high) // 2
        comparisons += 1
        if data[mid] == target:
            return mid, comparisons
        elif data[mid] < target:
            low = mid + 1
        else:
            high = mid - 1
    return -1, comparisons

index, comparisons = binary_search(data, target)

print("Target:", target)
print("Index:", index)
print("Comparisons:", comparisons)
```

```python
# [CELL 9] SEARCH COMPARISON
data = list(range(1, 101))
target = ((ID_SUM + SEED_DIGIT) % 100) + 1

li, lc = linear_search(data, target)
bi, bc = binary_search(data, target)

print("Target:", target)
print(f"Linear -> index={li}, comparisons={lc}")
print(f"Binary -> index={bi}, comparisons={bc}")
print("Reduction:", lc - bc)
```

```python
# [CELL 10] BUBBLE SORT
def bubble_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a)):
        swapped = False
        for j in range(len(a) - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                swaps += 1
                swapped = True
        if not swapped:
            break
    return a, comparisons, swaps

data = [
    (ID_SUM * 3) % 100, (SEED_DIGIT * 11) % 100,
    (NAME_LENGTH * 7) % 100, 42, 17, 89, 31, 56
]

sorted_data, comparisons, swaps = bubble_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 11] SELECTION SORT
def selection_sort(data):
    a, comparisons, swaps = data.copy(), 0, 0
    for i in range(len(a) - 1):
        minimum = i
        for j in range(i + 1, len(a)):
            comparisons += 1
            if a[j] < a[minimum]:
                minimum = j
        if minimum != i:
            a[i], a[minimum] = a[minimum], a[i]
            swaps += 1
    return a, comparisons, swaps

sorted_data, comparisons, swaps = selection_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Swaps:", swaps)
```

```python
# [CELL 12] INSERTION SORT
def insertion_sort(data):
    a, comparisons, shifts = data.copy(), 0, 0
    for i in range(1, len(a)):
        key, j = a[i], i - 1
        while j >= 0:
            comparisons += 1
            if a[j] <= key:
                break
            a[j + 1] = a[j]
            shifts += 1
            j -= 1
        a[j + 1] = key
    return a, comparisons, shifts

sorted_data, comparisons, shifts = insertion_sort(data)

print("Sorted:", sorted_data)
print("Comparisons:", comparisons)
print("Shifts:", shifts)
```

```python
# [CELL 13] EXECUTION-TIME MEASUREMENT
import time

def measure(sort_function, data, trials=5):
    times = []
    for _ in range(trials):
        start = time.perf_counter()
        result = sort_function(data)
        times.append(time.perf_counter() - start)
    return result, sum(times) / trials

data_size = 100 + ID_SUM * 10

performance_data = [
    (ID_SUM * 3 + i * 7) % 1000
    for i in range(data_size)
]

bubble_result, bubble_time = measure(bubble_sort, performance_data)
selection_result, selection_time = measure(selection_sort, performance_data)
insertion_result, insertion_time = measure(insertion_sort, performance_data)

print(f"Dataset Size: {data_size}")
print(f"Bubble Sort:    {bubble_time:.8f} s")
print(f"Selection Sort: {selection_time:.8f} s")
print(f"Insertion Sort: {insertion_time:.8f} s")
```

```python
# [CELL 14] SORTING PERFORMANCE COMPARISON
results = [
    ("Bubble Sort", bubble_result, bubble_time),
    ("Selection Sort", selection_result, selection_time),
    ("Insertion Sort", insertion_result, insertion_time)
]

print(f"{'Algorithm':<18}{'Comparisons':>12}{'Movements':>12}{'Time (s)':>12}")

for name, result, elapsed in results:
    print(
        f"{name:<18}{result[1]:>12}"
        f"{result[2]:>12}{elapsed:>12.8f}"
    )
```

```python
# --- STUDENT CONFIGURATION & EXERCISE 1 ---
LAST_NAME = "Dacua"
STUDENT_ID = "TUPM-26-3491"
FAVORITE_ARTIST = "mypm"

SEED_DIGIT = int(STUDENT_ID[-1])  # 1
ID_SUM = sum(int(d) for d in STUDENT_ID if d.isdigit())  # 25
NAME_LENGTH = len(LAST_NAME)  # 5
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(c.lower()) - ord('a') + 1 for c in FAVORITE_ARTIST if c.isalpha())

# 1. Generate Original Dataset (Sorted unique values)
original_dataset = [i * 3 for i in range(1, ID_SUM + 1)]

# 2. Rotate the Dataset
rotation_index = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_index:] + original_dataset[:rotation_index]

# 3. Target Incident
target_incident = ID_SUM + ARTIST_SUM

print("Original Dataset:", original_dataset)
print("Rotated Dataset:", rotated_dataset)
print("Target Incident:", target_incident)

# 4. Modified Binary Search for Rotated Sorted Array
def binary_search_rotated(arr, target):
    left, right = 0, len(arr) - 1
    comparisons = 0
    while left <= right:
        comparisons += 1
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid, comparisons
        if arr[left] <= arr[mid]:
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1
    return -1, comparisons

# 5. Linear Search
def linear_search(arr, target):
    comparisons = 0
    for i in range(len(arr)):
        comparisons += 1
        if arr[i] == target:
            return i, comparisons
    return -1, comparisons

rot_index, rot_comps = binary_search_rotated(rotated_dataset, target_incident)
lin_index, lin_comps = linear_search(rotated_dataset, target_incident)

print("\n--- RESULTS ---")
print(f"Rotated Binary Search: Index {rot_index}, Comparisons: {rot_comps}")
print(f"Linear Search: Index {lin_index}, Comparisons: {lin_comps}")
```

```python
# --- EXERCISE 2: University Enrollment Records ---

# 1. Generate Dataset A (Ascending values using SEED_DIGIT and ID_SUM)
dataset_a = sorted([i for i in range(SEED_DIGIT, SEED_DIGIT + ID_SUM * 2, 3)])

# 2. Generate Dataset B (Ascending values using ARTIST_LENGTH and NAME_LENGTH)
dataset_b = sorted([i for i in range(NAME_LENGTH, NAME_LENGTH + ARTIST_LENGTH * 15, 4)])

# 3. Consolidate them into one ordered collection using Two-Pointer Merge technique
def consolidate_datasets(arr1, arr2):
    merged = []
    comparisons = 0
    i = j = 0
    
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] < arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1
            
    # Append remaining elements
    while i < len(arr1):
        merged.append(arr1[i])
        i += 1
        
    while j < len(arr2):
        merged.append(arr2[j])
        j += 1
        
    return merged, comparisons

consolidated_dataset, total_comparisons = consolidate_datasets(dataset_a, dataset_b)

print("Dataset A:", dataset_a)
print("Dataset B:", dataset_b)
print("Consolidated Dataset:", consolidated_dataset)
print(f"Number of records before consolidation: Dataset A ({len(dataset_a)}), Dataset B ({len(dataset_b)})")
print(f"Number of records after consolidation: {len(consolidated_dataset)}")
print(f"Comparison Count: {total_comparisons}")

# Complexity Analysis
print("\n--- COMPLEXITY FOR EXERCISE 2 ---")
print("Time Complexity: O(n + m) where n and m are the lengths of the two datasets")
print("Space Complexity: O(n + m) for the newly merged collection")
```

```python
import time
import random

# --- EXERCISE 3: E-Commerce Order Processing ---

# 1. Generate Base Orders using student inputs (ID_SUM and ARTIST_SUM)
base_orders = [i for i in range(1, ID_SUM + 1)]
random.seed(SEED_DIGIT)
random.shuffle(base_orders)

# Three Input Conditions
cond_random = base_orders.copy()
cond_sorted = sorted(base_orders)
cond_reverse = sorted(base_orders, reverse=True)

conditions = {
    "Random": cond_random,
    "Sorted": cond_sorted,
    "Reverse-Ordered": cond_reverse
}

# Sorting Algorithms with tracking (Comparisons and Movements/Swaps)
def bubble_sort(arr):
    a = arr.copy()
    comparisons = 0
    movements = 0
    n = len(a)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if a[j] > a[j + 1]:
                a[j], a[j + 1] = a[j + 1], a[j]
                movements += 3 # 1 swap = 3 assignments
                swapped = True
        if not swapped:
            break
    return a, comparisons, movements

def selection_sort(arr):
    a = arr.copy()
    comparisons = 0
    movements = 0
    n = len(a)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if a[j] < a[min_idx]:
                min_idx = j
        if min_idx != i:
            a[i], a[min_idx] = a[min_idx], a[i]
            movements += 3
    return a, comparisons, movements

def insertion_sort(arr):
    a = arr.copy()
    comparisons = 0
    movements = 0
    for i in range(1, len(a)):
        key = a[i]
        movements += 1
        j = i - 1
        while j >= 0:
            comparisons += 1
            if a[j] > key:
                a[j + 1] = a[j]
                movements += 1
                j -= 1
            else:
                break
        a[j + 1] = key
        movements += 1
    return a, comparisons, movements

# Run and measure performance across conditions
print(f"{'Condition':<15} | {'Algorithm':<15} | {'Comparisons':<12} | {'Movements':<12} | {'Time (s)':<10}")
print("-" * 75)

for cond_name, data in conditions.items():
    for algo_name, algo_func in [("Bubble Sort", bubble_sort), ("Selection Sort", selection_sort), ("Insertion Sort", insertion_sort)]:
        start_time = time.perf_counter()
        sorted_arr, comps, movs = algo_func(data)
        end_time = time.perf_counter()
        elapsed = end_time - start_time
        
        # Verify sorting
        is_sorted = sorted_arr == sorted(data)
        
        print(f"{cond_name:<15} | {algo_name:<15} | {comps:<12} | {movs:<12} | {elapsed:<10.6f}")

print("\n--- BIG O & CASE ANALYSIS SUMMARY ---")
print("1. Bubble Sort: Best-case O(n) when sorted; Worst/Average O(n^2).")
print("2. Selection Sort: O(n^2) across all cases (independent of initial order for comparisons).")
print("3. Insertion Sort: Best-case O(n) when sorted; Worst/Average O(n^2) for reverse/random.")
```
